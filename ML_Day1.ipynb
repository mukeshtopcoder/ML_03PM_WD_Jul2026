### What is Machine Learning?
Machine Learning is a branch of AI (Artificial Intelligence)<br>
In ML computer learn patterns from the old/historical dataset

#### Traditional Programming
Program + Input => Output
#### Machine Learning Programming
Input + Output => Program (Rules - Learn Patterns) <br>
Program (Rules) + Input => Output (Predicted Output)<br>

### Types of ML
- Supervised Learning
- Unsupervised Learning
- Reinforcement Learning

1. Supervised Learning (Labeled)
    - In this learning in your dataset we have a output/target/labeled column
    - Classification
    - Regression
2. Unsupervised Learning (Unlabeled)
    - In this learning in our dataset there is no target/output/labeled column
    - Clustering
    - Dimension Reduction (PCA, tSNE , UMAP)
    - Asscociation Rules (Recommendation)
